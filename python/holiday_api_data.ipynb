```python
import requests
import pandas as pd

try:
    url = 'https://canada-holidays.ca/api/v1/provinces'
    response = requests.get(url)
    data = response.json()
    
    df = pd.json_normalize(
        data['provinces'],
        record_path='holidays',
        meta=['id', 'nameEn'],
        meta_prefix='province_'
    )
    
    df.to_csv(r"C:\Users\letov\Projects\hr_warehouse\holidays.csv", index=False)
    
    print('Data successfully saved')
    
except: 
    print('Issues saving data')
```
